```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image1=cv2.imread(r"C:\Users\Student\Desktop\pic\001.jpg",0 )
image2=cv2.imread(r"C:\Users\Student\Desktop\pic\02.jpg",0 )
```

```python
histogram1=cv2.calcHist([image1],[0],None,[256],[0,256])
histogram2=cv2.calcHist([image2],[0],None,[256],[0,256])
```

```python
histogram1_noramalized=histogram1.copy()
histogram2_noramalized=histogram2.copy()
```

```python
cv2.normalize(histogram1,histogram1_noramalized,alpha=0,beta=1,norm_type=cv2.NORM_MINMAX)
cv2.normalize(histogram2,histogram2_noramalized,alpha=0,beta=1,norm_type=cv2.NORM_MINMAX)
plt.figure(figsize=[10,7])
plt.subplot(221);plt.plot(histogram1);plt.title("bigger image before normalize")
plt.subplot(222);plt.plot(histogram2);plt.title("smaller image before normalize")
plt.subplot(223);plt.plot(histogram1_noramalized);plt.title("bigger image after normalize")
plt.subplot(224);plt.plot(histogram2_noramalized);plt.title("smaller image after normalize")
plt.show()
#between 0 and 1 minmax normalize
```
