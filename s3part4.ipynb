```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
import matplotlib.pyplot as plt
import seaborn as sb
from sklearn.metrics import confusion_matrix,classification_report
```

```python
actual=["cat","cat","cat","not cat","cat","not cat","not cat","cat","cat","not cat","cat","cat","cat","not cat","cat","not cat"]
predicted=["cat","not cat","cat","not cat","cat","not cat","cat","not cat","cat","not cat","cat","not cat","cat","not cat","cat","not cat"]
```

```python
cm=confusion_matrix(actual,predicted)
sb.heatmap(cm,annot=True,fmt="0.1f",xticklabels=["cat","not cat"],linewidth=3,linecolor="white",cmap="viridis")
plt.xlabel("predict",fontsize=13)
plt.ylabel("actual",fontsize=13)
plt.show()
#یکسری رو مثلا میگیم که خب حدس زده کت باشه ولی نبوده یا مثلا حدس زده و درست بوده و به این صورت چهار حالت میشه
```

```python
from sklearn import metrics
cm_display=metrics.ConfusionMatrixDisplay(confusion_matrix=cm,display_labels=["cat","not cat"])
cm_display.plot()
plt.xlabel("predict",fontsize=13)
plt.ylabel("actual",fontsize=13)
plt.show()
```

```python
print(classification_report(actual,predicted))
```
