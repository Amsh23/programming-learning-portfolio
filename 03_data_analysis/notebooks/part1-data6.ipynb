```python
import pandas as pd
import seaborn as sb
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
irisdata=iris.data
irisdata
```

```python
df=pd.DataFrame(irisdata,columns=iris.feature_names)
df.sample(10)
```

```python
iris.target_names
```

```python
iris.target
```

```python
df=pd.DataFrame(irisdata,columns=iris.feature_names)
df["target_names"] = iris.target
df
```

```python
df.info()
```

```python
df.describe()
```

df.hist(figsize=(10,8),bins=30)
plt.tight_layout()
plt.show()

```python
correlation = df.corr()
correlation
```

```python
plt.figure(figsize=(8,6))
sb.heatmap(correlation, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Matrix')
plt.show()
```

```python
sb.pairplot(df, hue='target_names')
plt.show()
```

```python
df.columns
```

```python
plt.figure(figsize=(10,6))
sb.scatterplot(x='sepal length (cm)', y='sepal width (cm)', hue='target_names', data=df, palette='deep')
plt.title('Sepal Length vs Sepal Width')
plt.show()
```
