```python
import matplotlib.pyplot as plt
```

```python
x=[5,7,8,10,3,17,2,99,11,12,9,6]
y=[99,86,87,88,99,45,103,87,78,68,60,80]
```

```python
plt.scatter(x,y)
plt.show()
```

```python
from scipy import stats
```

```python
slope,intercept,r,p,std_err=stats.linregress(x,y)
def myfunc(x):
    return slope*x+intercept
```

```python
mymodel=list(map(myfunc,x))
plt.scatter(x,y)
plt.plot(x,mymodel)
plt.show()
```
