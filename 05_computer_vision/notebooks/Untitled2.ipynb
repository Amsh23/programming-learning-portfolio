```python
import numpy as np
import matplotlib.pyplot as plt
import cv2
```

```python
image1=cv2.imread(r"C:\Users\Student\Desktop\pic08\sudoko.jpg")
gray=cv2.cvtColor(image1,cv2.COLOR_BGR2GRAY)
edge=cv2.Canny(gray,50,150)
lines=cv2.HoughLines(edge,1,np.pi/180,150)
for line in lines:
    rho,theta=line[0]
    a=np.cos(theta)
    b=np.sin(theta)
    x0=a*rho
    y0=b*rho
    x1=int(x0+1000*(-b))
    y1=int(y0+1000*(a))
    x2=int(x0-1000*(-b))
    y2=int(y0-1000*(a))
    cv2.line(image1,(x1,y1),(x2,y2),(0,0,255),2)
plt.imshow(image1[...,::-1])
plt.show()
```

```python
orginal_image=cv2.imread(r"C:\Users\Student\Desktop\pic08\highway.jpg")
img=orginal_image.copy()
blured=cv2.GaussianBlur(orginal_image,(5,5),0)
gray=cv2.cvtColor(blured,cv2.COLOR_BGR2GRAY)
edge=cv2.Canny(gray,50,200)
lines=cv2.HoughLinesP(edge,1,np.pi/180,50,minLineLength=20,maxLineGap=50)
for line in lines:
    x1,y1,x2,y2=line[0]
    cv2.line(img,(x1,y1),(x2,y2),(0,0,255),2)
plt.figure(figsize=[17,8])
plt.subplot(131);plt.imshow(orginal_image[...,::-1]);plt.title("orginal image")
plt.subplot(132);plt.imshow(edge,cmap="gray");plt.title("gray")
plt.subplot(133);plt.imshow(img,cmap="gray");plt.title("result")
plt.show()
```
