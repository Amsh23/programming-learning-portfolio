```python
import seaborn as sb
import matplotlib.pyplot as plt
import numpy as np
from scipy.stats import skew, kurtosis
np.random.seed(42)
datasets = {
    "Normal": np.random.normal(0,1,1000),
    "Positive Skew": np.random.exponential(scale=1, size=1000),
    "Negative Skew": -np.random.exponential(scale=1, size=1000),
    "High Kurtosis": np.random.laplace(0,1,1000),
    "Low Kurtosis": np.random.uniform(-2,2,1000)
}
fig, axes = plt.subplots(2,3,figsize=(15,8))
for ax, (title, data) in zip(axes.flatten(), datasets.items()):
    sb.histplot(data, bins=30, kde=True, ax=ax, color="skyblue", edgecolor="black")
    sk=skew(data)
    ku=kurtosis(data)
    ax.set_title(f"{title}\nSkew={sk:.2f}, Kurt={ku:.2f}")
    ax.axvline(np.mean(data), color="red", linestyle="--", label="Mean")
    ax.legend()
axes[1, 2].axis("off")
plt.tight_layout()
plt.show()
```
