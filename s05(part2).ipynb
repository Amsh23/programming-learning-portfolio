```python
sb.distplot(random.normal(loc=15,scale=2,size=10000))
plt.show()
```

```python
sb.distplot(random.normal(loc=50,scale=5,size=1000),label="normal", hist=False)
sb.distplot(random.binomial(n=100,p=0.5,size=1000),label="binomial",hist=False)
plt.legend()
plt.show()
```

```python
x=random.poisson(lam=2, size=10)
x
```

```python
sb.distplot(random.normal(loc=50,scale=7,size=1000),label="normal", hist=False)
sb.distplot(random.poisson(lam=50,size=10000),label="poisson",hist=False)
plt.legend()
plt.show()
```

```python
sb.distplot(random.binomial(n=1000,p=0.01,size=1000),label="binomial", hist=False)
sb.distplot(random.poisson(lam=10,size=1000),label="poisson",hist=False)
plt.legend()
plt.show()
```

```python
from scipy.stats import bernoulli
```

```python
data=bernoulli.rvs(size=1000, p=0.6) # rvs: random variables
ax=sb.distplot(data,color="red",hist_kws={"lw":25, "alpha":0.2})
ax.set(xlabel="bernouli", ylabel="frequency")
plt.show()
```
