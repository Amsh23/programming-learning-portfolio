```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
img = cv2.imread(r"C:\Users\Student\Desktop\pic08\s.png")
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blured = cv2.medianBlur(gray, 5)
rows = gray.shape[0]
circles = cv2.HoughCircles(gray, cv2.HOUGH_GRADIENT, 1, rows / 8, param1 = 200, param2 = 30)
if circles is not None:
    circles = np.uint16(np.around(circles))
    for c in circles[0, :]:
        center_x, center_y, radius = c
        center = (center_x, center_y)
        cv2.circle(img, center, radius, (255, 0, 255), 3)
plt.imshow(img[...,::-1])       
plt.show()
```

```python
img = cv2.imread(r"C:\Users\Student\Desktop\pic08\y2.jpg")
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
blured = cv2.medianBlur(gray, 5)
rows = gray.shape[0]
circles = cv2.HoughCircles(blured, cv2.HOUGH_GRADIENT, 1, rows / 8, param1 = 100, param2 = 30, minRadius = 50, maxRadius = 100)
if circles is not None:
    print(f"number of coins: {len(circles[0,:])}")
    circles = np.uint16(np.around(circles))
    for c in circles[0, :]:
        center_x, center_y, radius = c
        center = (center_x, center_y)
        cv2.circle(img, center, 1, (255, 0, 255), 3)
        cv2.circle(img, center, radius, (255, 0, 255), 3)
plt.imshow(img[...,::-1])       
plt.show()
```
