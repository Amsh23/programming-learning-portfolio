```python
import cv2
import matplotlib.pyplot as plt
import numpy as np
```

```python
src = cv2.imread(r"C:\Users\Student\Desktop\pic08\004.jpeg")
src = cv2.GaussianBlur(src,(3,3), 0)
src_gray = cv2.cvtColor(src, cv2.COLOR_BGR2GRAY)
dst = cv2.Laplacian(src_gray, cv2.CV_16S, ksize = 3)
abs_dst = cv2.convertScaleAbs(dst)
plt.figure(figsize = (12, 7))
plt.subplot(121);plt.imshow(src[...,::-1]);plt.title("original")
plt.subplot(122);plt.imshow(abs_dst, cmap = "gray");plt.title("Laplacian")
plt.show()
```

```python
src = cv2.imread(r"C:\Users\Student\Desktop\pic08\004.jpeg")
edge = cv2.Canny(src, 100, 200)
plt.figure(figsize = (12, 7))
plt.subplot(121);plt.imshow(src[...,::-1]);plt.title("original")
plt.subplot(122);plt.imshow(edge, cmap = "gray");plt.title("Canny")
plt.show()
```

```python
cap = cv2.VideoCapture(0)
while True:
    ret, frame = cap.read()
    src_gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
    edge = cv2.Canny(src_gray, 100, 200)
    edge = cv2.bitwise_not(edge)
    cv2.imshow("webcam", edge)
    if cv2.waitKey(1)&0xff == 27:
        break
cv2.release()
cv2.destroyAllWindows()
```
