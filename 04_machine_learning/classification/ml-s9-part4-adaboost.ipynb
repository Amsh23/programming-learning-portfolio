```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,classification_report
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
x,y=iris.data,iris.target
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
base_model=DecisionTreeClassifier(max_depth=2)
model=AdaBoostClassifier(base_estimator=base_model,n_estimators=50,learning_rate=1.0,random_state=42)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print(f"accuracy_score={accuracy_score(y_test,y_pred)}")
```

```python
print(f"classification_report={classification_report(y_test,y_pred)}")
```

```python
from sklearn.ensemble import AdaBoostRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error,r2_score
#from sklearn.datasets import fetch_california_housing
#نت اوکی شدش اینو تست کنم
import numpy as np
import pandas as pd
```

```python
data=pd.read_csv(r"C:\Users\Student\Desktop\Boston.csv")
```

```python
data.columns
```

```python
x=data[['crim', 'zn', 'indus', 'chas', 'nox', 'rm', 'age', 'dis',
       'rad', 'tax', 'ptratio', 'black', 'lstat']].values
y=data["medv"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
base=DecisionTreeRegressor(max_depth=4)
model=AdaBoostRegressor(base_estimator=base,n_estimators=100,learning_rate=0.1,random_state=42)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
print(f"mse={round(mean_squared_error(y_test,y_pred),2)}")
print(f"rmse={round(np.sqrt(mean_squared_error(y_test,y_pred)),2)}")
print(f"r2score={round(r2_score(y_test,y_pred),2)}")
```
