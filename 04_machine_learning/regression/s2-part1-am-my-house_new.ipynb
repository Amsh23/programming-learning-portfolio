```python
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings("ignore")
```

```python
data_house=pd.read_csv(r"C:\Users\Student\Desktop\house.csv")
data_house.sample(5)
```

```python
data_house.dtypes
```

```python
data_house["Area"]=pd.to_numeric(data_house["Area"],errors="coerce")
```

```python
data_house.replace({True:1,False:0},inplace=True)
data_house.sample(5)
```

```python
data_house.isnull().sum()
```

```python
data_house.dropna(subset=["Area","Address"],inplace=True)
```

```python
data_house.isnull().sum()
```

```python
data_house.drop_duplicates(inplace=True)
```

```python
data_house.reset_index(inplace=True,drop=True)
```

```python
address_map={adre:i for i, adre in enumerate(data_house["Address"].unique(),start=1)}
data_house["Address"]=data_house["Address"].map(address_map)
```

```python
data_house.sample(5)
```

```python
address_avg_price=data_house.groupby("Address").apply(lambda group:group["Price"].sum()/group["Area"].sum())
data_house["Average_mean_price"]=data_house["Address"].map(address_avg_price)
data_house.sample(3)
```

```python
data_house.columns
```

```python
feature=['Area', 'Room', 'Parking', 'Warehouse', 'Elevator', 'Address']
x=data_house[feature].values
y=data_house["Price"].values
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=4)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
from sklearn.linear_model import LinearRegression
```

```python
reg=LinearRegression()
reg.fit(x_train,y_train)
```

```python
y_pred=reg.predict(x_test)
```

```python
from sklearn.linear_model import Ridge
from sklearn.metrics import mean_squared_error
```

```python
ridge=Ridge(alpha=0.1)
ridge.fit(x_train,y_train)
```

```python
y_ridge=ridge.predict(x_test)
ridge_mse=mean_squared_error(y_test,y_ridge)
ridge_mse
```

```python
np.round(ridge_mse,2)
```

```python
round(reg.score(x_train,y_train),2)
```

```python
round(reg.score(x_test,y_test),2)
```

```python
ridge_score_train=ridge.score(x_train,y_train)
round(ridge_score_train,2)
```

```python
ridge_score_train=ridge.score(x_test,y_test)
round(ridge_score_train,2)
```
