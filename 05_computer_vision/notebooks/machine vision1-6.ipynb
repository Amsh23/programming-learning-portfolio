```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread(r"C:/Users/Student/Desktop/pic/v.jpg")
```

```python
hsvimg=cv2.cvtColor(img,cv2.COLOR_BGR2HSV)
```

```python
cv2.imshow("HSV image",hsvimg)
cv2.imshow("hue channel",hsvimg[:,:,-90])
cv2.imshow("saturation channel",hsvimg[:,:,1])
cv2.imshow("value channel",hsvimg[:,:,2])
cv2.waitKey()
cv2.destroyAllWindows()
```

```python
plt.figure(figsize=[20,5])
plt.subplot(131);plt.imshow(hsvimg[:,:,0],cmap="gray");plt.title("hue");
plt.subplot(132);plt.imshow(hsvimg[:,:,1],cmap="gray");plt.title("saturatin");
plt.subplot(133);plt.imshow(hsvimg[:,:,2],cmap="gray");plt.title("value");
plt.show()
```

```python
saturated_img=cv2.merge([hsvimg[:,:,0],cv2.add(hsvimg[...,1],50),hsvimg[...,2]])
saturated_img=cv2.cvtColor(saturated_img,cv2.COLOR_HSV2BGR)
plt.figure(figsize=[12,5])
plt.subplot(121);plt.imshow(saturated_img[...,::-1]);plt.title("saturated_image");
plt.subplot(122);plt.imshow(img[...,::-1]);plt.title("original");
plt.show()
```

```python
bright_img=cv2.merge([hsvimg[:,:,0],hsvimg[...,1],cv2.add(hsvimg[...,2],100)])
bright_img=cv2.cvtColor(bright_img,cv2.COLOR_HSV2BGR)
plt.figure(figsize=[12,5])
plt.subplot(121);plt.imshow(bright_img[...,::-1]);plt.title("bright_image");
plt.subplot(122);plt.imshow(img[...,::-1]);plt.title("original");
plt.show()
```
