```python
import cv2,matplotlib.pyplot as plt,numpy as np
```

```python
list_of_number=np.array([5,4,5,10,3,3,6,9,7,8,7,9,1,3,7])
plt.figure(figsize=[10,3])
plt.subplot(121);plt.hist(list_of_number);plt.title("histogram")
plt.subplot(122);plt.hist(list_of_number,5,color="pink");plt.title("histogram with 5 bins")
plt.show()
```

```python
image=cv2.imread(r"C:\Users\Student\Desktop\pic06\02.jpg",0)
plt.figure(figsize=[10,3])
plt.subplot(121);plt.hist(image.ravel(),256,color="orange");plt.title("histogram")
plt.subplot(122);plt.hist(image.ravel(),256,color="magenta",histtype="step");plt.title("histogram with step")
plt.show()
```

```python
histogram=cv2.calcHist([image],[0],None,[256],[0,256])
plt.figure(figsize=[5,3])
plt.plot(histogram)
plt.show()
```

```python
dark_ton=cv2.imread(r"C:\Users\Student\Desktop\dark.png")
mid_ton=cv2.imread(r"C:\Users\Student\Desktop\mid.png")
light_ton=cv2.imread(r"C:\Users\Student\Desktop\light.png")
dark_to_gray=cv2.cvtColor(dark_ton,cv2.COLOR_BGR2GRAY)
mid_to_gray=cv2.cvtColor(mid_ton,cv2.COLOR_BGR2GRAY)
light_to_gray=cv2.cvtColor(light_ton,cv2.COLOR_BGR2GRAY)
dark_ton_histogram=cv2.calcHist([dark_to_gray],[0],None,[256],[0,256])
mid_ton_histogram=cv2.calcHist([mid_to_gray],[0],None,[256],[0,256])
light_ton_histogram=cv2.calcHist([light_to_gray],[0],None,[256],[0,256])
plt.figure(figsize=[12,7])
plt.subplot(231);plt.imshow(dark_to_gray,cmap="gray");plt.title("dark_to_gray")
plt.subplot(232);plt.imshow(mid_to_gray,cmap="gray");plt.title("mid_to_gray")
plt.subplot(233);plt.imshow(light_to_gray,cmap="gray");plt.title("light_to_gray")

plt.subplot(234);plt.plot(dark_ton_histogram,cmap="gray");plt.title("dark_ton_histogram")
plt.subplot(235);plt.plot(mid_ton_histogram,cmap="gray");plt.title("mid_ton_histogram")
plt.subplot(236);plt.plot(light_ton_histogram,cmap="gray");plt.title("light_ton_histogram")
plt.show()
```
