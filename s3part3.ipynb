```python
import numpy as np
import pandas as pd
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import train_test_split,GridSearchCV
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
df["restecg"].value_counts()
```

```python
df=pd.get_dummies(df,columns=['restecg'])
df
```

```python
label=[]
for i in df.columns:
    if i!="output":
        label.append(i)
x=np.asarray(df[label])
y=np.asarray(df["output"])
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
s=StandardScaler()
s.fit_transform(x)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.3,random_state=42,stratify=df["sex"])
```

```python
scaler=StandardScaler()
x_train_scale=scaler.fit_transform(x_train)
x_test_scale=scaler.transform(x_test)
```

```python
p_grid={"n_neighbors":np.arange(1,50)}
knn=KNeighborsClassifier()
knn_cv=GridSearchCV(knn,p_grid,cv=5,scoring="accuracy")
knn_cv.fit(x_train_scale,y_train)
print(knn_cv.best_params_)
print(knn_cv.best_score_)
```

```python
knn=KNeighborsClassifier(n_neighbors=10,metric="minkowski",p=2)
knn.fit(x_train_scale,y_train)
knn.score(x_test_scale,y_test)
```
