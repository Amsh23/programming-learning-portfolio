```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
np.random.seed(4)
ages = np.random.randint(20,65,size=15)
salarie=np.random.randint(1000,10000,size=15)
df=pd.DataFrame({"Age":ages, "Salary":salarie})
df.head()
```

```python
from sklearn.preprocessing import Normalizer
```

```python
normalizer=Normalizer()
normalizerdata=normalizer.fit_transform(df)
normalizerdata
```

```python
normalizerdf=pd.DataFrame(normalizerdata,columns=df.columns,index=df.index)
normalizerdf.head()
```

```python
plt.plot(normalizerdf.index,normalizerdf["Age"])
plt.plot(normalizerdf.index,normalizerdf["Salary"])
plt.show()
```

```python
from sklearn.preprocessing import RobustScaler
```

```python
robust=RobustScaler()
robustdata=robust.fit_transform(df)
robustdata
```

```python
robustdf=pd.DataFrame(robustdata,columns=df.columns,index=df.index)
robustdf.head()
```

```python
plt.plot(robustdf.index,robustdf["Age"], label="age")
plt.plot(robustdf.index,robustdf["Salary"],label="salary")
plt.legend()
plt.show()
```

```python
from sklearn.preprocessing import MaxAbsScaler
```

```python
maxabs=MaxAbsScaler()
maxabsdata=maxabs.fit_transform(df)
maxabsdata
```

```python
maxabsdf=pd.DataFrame(maxabsdata,columns=df.columns,index=df.index)
maxabsdf.head()
```

```python
plt.plot(maxabsdf.index,maxabsdf["Age"], label="age")
plt.plot(maxabsdf.index,maxabsdf["Salary"],label="salary")
plt.legend()
plt.savefig(r"C:\Users\Student\Desktop\sara.png")
plt.show()
```
