```python
from sklearn.cluster import MeanShift
from sklearn.datasets import make_blobs
import matplotlib.pyplot as plt
```

```python
x,y=make_blobs(n_samples=100,centers=3,random_state=42,cluster_std=0.9)
```

```python
meanshift=MeanShift()
meanshift.fit(x)
```

```python
meanshift.labels_
```

```python
set(meanshift.labels_)
```

```python
meanshift.cluster_centers_
```

```python
plt.scatter(x[:,0],x[:,1],c=meanshift.labels_,cmap="Set2",marker="o")
plt.scatter(meanshift.cluster_centers_[:,0],meanshift.cluster_centers_[:,1],marker="x",s=100,linewidth=3)
plt.show()
```
