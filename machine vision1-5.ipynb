```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img=cv2.imread(r"C:/Users/Student/Desktop/pic/x.jpg")
b,g,r=cv2.split(img)
```

```python
modify_b=b+30
```

```python
merge=cv2.merge([modify_b,g,r])
```

```python
plt.figure(figsize=[20,5])
plt.subplot(151);plt.imshow(r,cmap="gray");plt.title("red_channel");
plt.subplot(152);plt.imshow(g,cmap="gray");plt.title("green_channel");
plt.subplot(153);plt.imshow(modify_b,cmap="gray");plt.title("modify_b channel");
plt.subplot(154);plt.imshow(merge[...,::-1]);plt.title("merge");
plt.subplot(155);plt.imshow(img[...,::-1]);plt.title("original");
plt.show()
#OVERFLOW
```
