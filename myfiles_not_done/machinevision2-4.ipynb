```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image=cv2.imread(r"C:\Users\Student\Desktop\pic2\000.jpg")
```

```python
gamma=0.4
lookuptable=np.empty((1,256),np.uint8)
for i in range(256):
    lookuptable[0,i]=np.clip(pow(i/255,gamma)*255,0,255)
out=cv2.LUT(image,lookuptable)
plt.figure(figsize=[15,5])
plt.subplot(121);plt.imshow(image[...,::-1]);plt.title("original")
plt.subplot(122);plt.imshow(out[...,::-1]);plt.title("new_image")
plt.show()
```

```python
gamma=0.4
image1=cv2.imread(r"C:\Users\Student\Desktop\pic2\a.jpg")
image2=cv2.imread(r"C:\Users\Student\Desktop\pic2\aaa.jpg")
output_linear1=cv2.convertScaleAbs(image1,alpha=1.3,beta=40)
output_linear2=cv2.convertScaleAbs(image2,alpha=1.3,beta=40)
for i in range(256):
    lookuptable[0,i]=np.clip(pow(i/255,gamma)*255,0,255)
output1_gamma=cv2.LUT(image1,lookuptable)
output2_gamma=cv2.LUT(image2,lookuptable)
plt.figure(figsize=[15,5])
plt.subplot(231);plt.imshow(image1[...,::-1]);plt.title("original")
plt.subplot(232);plt.imshow(output_linear1[...,::-1]);plt.title("linear")
plt.subplot(233);plt.imshow(output1_gamma[...,::-1]);plt.title("gamma")
plt.subplot(234);plt.imshow(image2[...,::-1]);plt.title("original")
plt.subplot(235);plt.imshow(output_linear2[...,::-1]);plt.title("linear")
plt.subplot(236);plt.imshow(output2_gamma[...,::-1]);plt.title("gamma")
plt.show()
```
