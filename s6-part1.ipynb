```python
import cv2,matplotlib.pyplot as plt,numpy as np
```

```python
bright=cv2.imread(r"C:\Users\Student\Desktop\pic06\cube-bright.png")
dark=cv2.imread(r"C:\Users\Student\Desktop\pic06\cube-dark.png")
```

```python
plt.figure(figsize=[11,11])
plt.subplot(331);plt.imshow(bright[...,::-1]);plt.title("bright")
plt.subplot(332);plt.imshow(dark[...,::-1]);plt.title("dark")
plt.subplot(334);plt.imshow(bright[...,0],cmap="gray");plt.title("bright-blue")
plt.subplot(335);plt.imshow(bright[...,1],cmap="gray");plt.title("bright-green")
plt.subplot(336);plt.imshow(bright[...,2],cmap="gray");plt.title("bright-red")
plt.subplot(337);plt.imshow(dark[...,0],cmap="gray");plt.title("dark-blue")
plt.subplot(338);plt.imshow(dark[...,1],cmap="gray");plt.title("dark-green")
plt.subplot(338);plt.imshow(dark[...,2],cmap="gray");plt.title("dark-red")
plt.show()
```

```python
brighthsv=cv2.cvtColor(bright,cv2.COLOR_BGR2HSV)
darkhsv=cv2.cvtColor(dark,cv2.COLOR_BGR2HSV)
plt.figure(figsize=[11,11])
plt.subplot(331);plt.imshow(bright[...,::-1]);plt.title("bright")
plt.subplot(332);plt.imshow(dark[...,::-1]);plt.title("dark")
plt.subplot(334);plt.imshow(brighthsv[...,0],cmap="gray");plt.title("brighthsv-blue")
plt.subplot(335);plt.imshow(brighthsv[...,1],cmap="gray");plt.title("brighthsv-green")
plt.subplot(336);plt.imshow(brighthsv[...,2],cmap="gray");plt.title("brighthsv-red")
plt.subplot(337);plt.imshow(darkhsv[...,0],cmap="gray");plt.title("darkhsv-blue")
plt.subplot(338);plt.imshow(darkhsv[...,1],cmap="gray");plt.title("darkhsv-green")
plt.subplot(338);plt.imshow(darkhsv[...,2],cmap="gray");plt.title("darkhsv-red")
plt.show()
```
