```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from numpy import random
```

```python
random.seed(4)
ages=random.randint(20,60,size=20)
salaries=random.randint(10000,100000,size=20)
```

```python
df=pd.DataFrame({"Age":ages, "Salary":salaries})
df.head()
```

```python
plt.plot(df.index,df["Age"], color="deeppink")
plt.plot(df.index,df["Salary"], color="cyan")
plt.show()
```

<h1> test </h1>

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
minmax=MinMaxScaler()
minmaxdata=minmax.fit_transform(df)
minmaxdata
```

```python
minmaxdf = pd.DataFrame(minmaxdata,index=df.index,columns=df.columns)
minmaxdf.head()
```

```python
plt.plot(df.index,minmaxdf["Age"], color="deeppink")
plt.plot(df.index,minmaxdf["Salary"], color="cyan")
plt.xticks(np.arange(0,20,1))
plt.show()
```
