```python
import numpy as np
import pandas as pd
```

```python
movies_df=pd.read_csv(r"C:\Users\Student\Desktop\movies.csv")
ratings_df=pd.read_csv(r"C:\Users\Student\Desktop\ratings.csv")
```

```python
movies_df.head()
```

```python
movies_df["year"]=movies_df.title.str.extract("(\(\d\d\d\d\))") #digits or (r"(\(\d{4}\))")
movies_df["year"]=movies_df.year.str.extract("(\d\d\d\d)") #or (r"(\d{4})")
movies_df["title"]=movies_df.title.str.replace("(\(\d\d\d\d\))","",regex=True) #tonone
movies_df["title"]=movies_df["title"].apply(lambda x:x.strip()) #removespacesifexist
```

```python
movies_df.head()
```

```python
movies_df["genres"]=movies_df.genres.str.split("|")
movies_df.head()
```

```python
moviewithgenres_df=movies_df.copy()
for index,row in movies_df.iterrows():
    for genre in row["genres"]:
        moviewithgenres_df.at[index,genre]=1
        
moviewithgenres_df=moviewithgenres_df.fillna(0)
moviewithgenres_df.head()
```

```python
ratings_df.head()
```

```python
ratings_df=ratings_df.drop("timestamp",axis=1)
ratings_df.head()
```

```python
userinput=[{"title":"Toy Story","rating":3.5},
          {"title":"Dracula: Dead and Loving It","rating":5},
          {"title":"Jumanji","rating":2},
          {"title":"Casino","rating":5},
          {"title":"Copycat","rating":3}]
```

```python
inputmovies=pd.DataFrame(userinput)
inputmovies
```

```python
inputId=movies_df[movies_df["title"].isin(inputmovies["title"].tolist())]
inputmovies=pd.merge(inputId,inputmovies,on="title",how="inner") #داخلی و روی
inputmovies
```

```python
inputmovies=inputmovies.drop(columns=["genres","year"])
```

```python
usermovies=moviewithgenres_df[moviewithgenres_df["movieId"].isin(inputmovies["movieId"].tolist())]
```

```python
usermovies
```

```python
usermovies=usermovies.reset_index(drop=True)
userGenreTable=usermovies.drop(columns=["movieId","title","genres","year"])
userGenreTable
```

```python
inputmovies["rating"]
```

```python
userprofile=userGenreTable.transpose().dot(inputmovies["rating"])
userprofile
```

```python
genreTable=moviewithgenres_df.set_index(moviewithgenres_df["movieId"])
genreTable=genreTable(columns=["movieId","title","genres","year"])
genreTable.head()
#چک کنم با فایل یا ویدیو کلاس
```

```python
recomendationTable_df=((genreTable*userprofile).sum(axis=1)/(userprofile.sum())
recomendationTable_df.head()
```

```python
recomendationTable_df=recomendationTable_df.sort_values(ascendinf=False)
recomendationTable_df.head()
```
