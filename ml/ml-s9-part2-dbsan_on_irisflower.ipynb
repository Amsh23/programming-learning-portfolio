```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import DBSCAN
from sklearn.datasets import load_iris
```

```python
iris=load_iris()
x=iris.data
```

```python
dbscan=DBSCAN(eps=0.5,min_samples=4)
dbscan.fit(x)
label=dbscan.labels_
```

```python
set(label)
#ست آیتم های تکراری را حذف میکرد
```

```python
np.histogram(label)
```

```python
plt.hist(label)
plt.show()
```

```python
x=pd.DataFrame(x)
x["target"]=list(dbscan.labels_)
x.head()
```

```python
x=x[x["target"]!=-1]
x.shape
```

```python
y=x["target"]
```

```python
x=x.values
y=y.values
```

```python
colors=["m","y","b","g"]
for i in range(len(x)):
    plt.scatter(x[i,0],x[i,1],c=colors[label[i]])
    plt.annotate(str(label[i]),(x[i,0],x[i,1]))
plt.show()
```

```python
labels=dbscan.fit_predict(x)
pd.Series(labels).value_counts()
```
