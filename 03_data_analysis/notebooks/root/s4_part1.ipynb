```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sb
```

```python
insurance_data = pd.read_csv(r"C:\Users\Student\Desktop\insurance.csv")
insurance_data
```

```python
insurance_data.sample(3)
```

```python
sb.scatterplot(x=insurance_data["bmi"],y=insurance_data["charges"])
plt.show()
```

```python
sb.scatterplot(x=insurance_data["bmi"],y=insurance_data["charges"],hue=insurance_data["smoker"])
plt.show()
```

```python
sb.scatterplot(x=insurance_data["bmi"],y=insurance_data["charges"],hue=insurance_data["smoker"], palette="Set2")
plt.show()
```

```python
sb.regplot(x=insurance_data["bmi"],y=insurance_data["charges"])
plt.show()
```

```python
sb.lmplot(x="bmi",y="charges",hue="smoker",data=insurance_data)
plt.show()
```

```python
sb.histplot(insurance_data["age"], bins="auto", kde=True)
plt.xticks(np.arange(18,66,2),rotation=-90)
plt.yticks(np.arange(0,200,10))

plt.show()
```

```python
sb.boxplot(x="sex",y="bmi",data=insurance_data)
plt.show()
```

```python
sb.countplot(x="smoker", data=insurance_data, palette="viridis")
plt.show()
```
