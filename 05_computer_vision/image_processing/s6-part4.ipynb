```python
import cv2,matplotlib.pyplot as plt,numpy as np
```

```python
image=cv2.imread(r"C:\Users\Student\Desktop\pic06\xray.jpg",0)
result=cv2.equalizeHist(image)
plt.figure(figsize=[12,7])
plt.subplot(221);plt.imshow(image,cmap="gray");plt.title("original")
plt.subplot(222);plt.hist(image.ravel(),256)
plt.subplot(223);plt.imshow(result,cmap="gray");plt.title("equalize hist")
plt.subplot(224);plt.hist(result.ravel(),256)
plt.show()
```
