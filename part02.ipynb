```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x=np.array([0,10])
y=np.array([10,100])
```

```python
font1={"family":"serif", "color":"blue","size":20}
font2={"family":"sans serif", "color":"red","size":20}
```

```python
plt.plot(x,y,"b", ms=20,mec="g", mfc="y",lw=3)
plt.title("chart1",loc="left")
plt.xlabel("axis x", fontdict=font1)
plt.ylabel("axis y", fontdict=font2)
plt.show()
```

```python
lesson=np.array(["python","ml", "nlp", "deeplearning", "java script"])
mft=np.array([1000,850,500,380,220])
sharif=np.array([800,620,400,300,210])
```

```python
# lesson=np.array(["p","ml","nlp","data","java"])
# number=np.array([1000,850,500,380,220])

plt.plot(lesson,mft, sharif)
plt.title("chart1",loc="left")
plt.xlabel("lesson", fontdict=font1)
plt.ylabel("number", fontdict=font2)
plt.show()
```

```python
plt.xlabel("lesson")
plt.ylabel("number")
plt.xticks(rotation=90)
plt.plot(lesson,mft,ls="--", marker="*", mew=3, color="silver")
plt.plot(lesson,sharif,ls=":", marker="+", mew=3, color="magenta")
plt.legend(["mft","sharif"], loc="best")
plt.show()

```

```python
plt.xlabel("lesson")
plt.ylabel("number")
plt.xticks(rotation=90)
plt.plot(lesson,mft,ls="--", marker="*", mew=3, color="silver")
plt.plot(lesson,sharif,ls=":", marker="+", mew=3, color="magenta")
plt.legend(["mft","sharif"], loc="best")
plt.annotate("mftpython", xytext=(0.5,800), xy=(0,1000), arrowprops={"facecolor":"cyan", "width":2}, fontsize=10)
plt.show()

```

```python
plt.xlabel("lesson")
plt.ylabel("number")
plt.plot(lesson,number)

for i in range(len(lesson)):
    plt.text(i,number[i],lesson[i])

plt.show()
```
